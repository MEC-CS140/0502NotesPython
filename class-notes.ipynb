# Functions

```python
def my_function():
    print("this is a print statement inside a function")

my_function()

def function_return():
    # a way to create variables on one line
    num1, num2 = 1, 5
    product = num1 * num2
    return product
    print("this code is never reached")

function_output = function_return()
print(function_output)

def function_args(num1, num2):
    product = num1 * num2
    return product

function_output = function_args(10, 15)
print(function_output)

def get_output(value_to_print):
    print(value_to_print)

def my_new_function():
    value = function_args(10, 20)
    get_output(value)

my_new_function()
```

```python
def convert_fahrenheit(temp_celsius):
    temp_f = (temp_celsius - 32) * 5/9
    return temp_f  

print(convert_fahrenheit(32))

print(convert_fahrenheit(int(input("Enter temp in celsius"))))

```

# Lists

```python
colors = ["red", "yellow", "green", "blue"]
print(colors)

colors.append("orange")
print(colors)

colors.append("test")
print(colors)

colors.remove("test")
print(colors)

colors.pop()
print(colors)

def print_list(list, command, value = None):
    
    if command == "append" and value is not None:
        list.append(value)

    if command == "remove" and value is not None:
        list.remove(value)
    
    if command == "pop":
        list.pop()
    
    if command == "exists" and value is not None:
        print(value in list)

    print(list)


print_list(colors, "print")

print_list(colors, "append", "orange")

print_list(colors, "append", "test")
print_list(colors, "remove", "test")
print_list(colors, "pop")



new_list = []
print_list(new_list, "append", "new value")
```
